On dispose au début de chaque épisode de $N_0$ unités d'un bien qu'on souhaite vendre au fil des étapes. À l'étape $t\geqslant 0$, il nous en reste une quantité $N_t$. L'offre $(Q_t,P_t)$ est tirée selon une distribution de probabilité (éventuellement inconnue) dépendant de $N_t$, qu'on notera $\nu_{N_t}$.
